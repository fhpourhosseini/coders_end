```python
import json
import pandas as pd

try:
    # Load the raw data you captured from your Web Developer Tools
    with open("raw_fire_data.json", "r", encoding="utf-8") as f:
        raw_data = json.load(f)

    # Convert the server arrays instantly into a clean spreadsheet structure
    df = pd.DataFrame(raw_data)

    output_filename = "boku_fires_2002_2026_clean.csv"
    df.to_csv(output_filename, index=False, encoding="utf-8-sig")

    print(f"Success! Processed {len(df)} forest fire records.")
    print(f"Spreadsheet saved to: {output_filename}")

except FileNotFoundError:
    print(
        "Missing input file. Please save your copied network response as 'raw_fire_data.json' first."
    )
except json.JSONDecodeError:
    print(
        "The text inside 'raw_fire_data.json' doesn't look like a valid JSON payload. Double-check your copy."
    )
except Exception as e:
    print(f"Error during processing: {e}")
```
