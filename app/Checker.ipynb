```python
import json
import pandas as pd

# 1. Count features in the original JSON
with open("../data/raw_fire_data.json", "r", encoding="utf-8") as f:
    json_data = json.load(f)
json_count = len(json_data.get("features", []))

# 2. Count rows in the generated CSV
df = pd.DataFrame(pd.read_csv("../data/boku_waldbrand_02-11-2026.csv"))
csv_count = len(df)

print(f"Total entries in original JSON: {json_count}")
print(f"Total entries in generated CSV:  {csv_count}")

if json_count == csv_count:
    print("Verification Passed: Every single fire event was successfully migrated!")
else:
    print("Verification Failed: The row counts do not match.")
```
